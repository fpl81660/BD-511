# Actividad 3 - Estadística descriptiva

```python
F_pais =df.loc[(df["Entity"] == "Portugal") & (df["Year"] >= 2006) & (df["Year"] <= 2020), "Annual change in population" ]

print("Media:", F_pais.mean())
print("Mediana:", F_pais.median())
print("Desviación Estándar:", F_pais.std())
print("Mínimo:", F_pais.min())
print("Máximo:", F_pais.max())
```
